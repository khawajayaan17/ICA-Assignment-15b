```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("C:/Users/Hp/Downloads/ride_hailing_dataset.csv")

# Q1: Average fare by service type
fare_by_service = (
    df.groupby("service_type")["fare"]
    .agg(["count", "mean", "median", "std"])
    .sort_values("mean", ascending=False)
)
print("Q1 - Average Fare by Service Type:\n", fare_by_service)

# Q2: City with the most trips
trips_by_city = df["city"].value_counts()
print("\nQ2 - Trips by City:\n", trips_by_city)
print("City with most trips:", trips_by_city.idxmax())

# Q3: Average distance per trip
avg_distance = df["distance_km"].mean()
print(f"\nQ3 - Average distance per trip: {avg_distance:.2f} km")

# Q4: Compare trip ratings by gender
ratings_by_gender = df.groupby("gender")["trip_rating"].agg(
    ["count", "mean", "median", "std"]
)
print("\nQ4 - Trip Ratings by Gender:\n", ratings_by_gender)

# Q5: Bar chart of service type vs. average fare
q5_data = (
    df.groupby("service_type")["fare"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(8, 5))
sns.barplot(data=q5_data, x="service_type", y="fare", palette="viridis")
plt.title("Average Fare by Service Type", fontsize=14)
plt.xlabel("Service Type", fontsize=12)
plt.ylabel("Average Fare", fontsize=12)
plt.tight_layout()
plt.show()
```
